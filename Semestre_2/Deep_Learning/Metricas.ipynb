```python
import numpy as np
```

```python
cm = np.array([
    [7, 8, 9],
    [1, 2, 3],
    [3, 2, 1]
])

total = cm.sum()
print("Total:", total)
```

![Matriz de Confusion](img/cm.png)

# Manzana

```python
TP = 7
FP = 1 + 3  # Otras frutas clasificadas como manzana
FN = 8 + 9  # Manzanas mal clasificadas
TN = total - (TP + FP + FN)

precision = TP / (TP + FP)
recall = TP / (TP + FN)
specificity = TN / (TN + FP)
f1 = 2 * precision * recall / (precision + recall)

print("TP:", TP, "FP:", FP, "FN:", FN, "TN:", TN)
print("Precisión:", round(precision,3))
print("Recall:", round(recall,3))
print("Especificidad:", round(specificity,3))
print("F1:", round(f1,3))
```

# Naranja

```python
TP = 2
FP = 8 + 2   # Predichas Naranja pero eran otras
FN = 1 + 3   # Naranjas mal clasificadas
TN = total - (TP + FP + FN)

precision = TP / (TP + FP)
recall = TP / (TP + FN)
specificity = TN / (TN + FP)
f1 = 2 * precision * recall / (precision + recall)

print("TP:", TP, "FP:", FP, "FN:", FN, "TN:", TN)
print("Precisión:", round(precision,3))
print("Recall:", round(recall,3))
print("Especificidad:", round(specificity,3))
print("F1:", round(f1,3))
```
