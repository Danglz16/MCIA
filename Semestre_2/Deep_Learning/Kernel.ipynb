# Cargar Librerias

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
RUTA = "img/doggo.png"

img_bgr = cv2.imread(RUTA, cv2.IMREAD_COLOR)
img32 = img_bgr.astype(np.float32)

K1 = np.array([[0, 0, 0],
               [0, 1, 0],
               [0, 0, 0]], dtype=np.float32)

K2 = np.array([[ 1, 0,-1],
               [ 0, 1, 0],
               [-1, 0, 1]], dtype=np.float32)

K3 = np.array([[ 0,-1, 0],
               [-1, 4,-1],
               [ 0,-1, 0]], dtype=np.float32)  

K4 = np.array([[-1,-1,-1],
               [-1, 8,-1],
               [-1,-1,-1]], dtype=np.float32)

K5 = np.array([[ 0,-1, 0],
               [-1, 5,-1],
               [ 0,-1, 0]], dtype=np.float32)

K6_M = np.array([[ 1, 2, 1],
               [ 2, 4, 2],
               [ 1, 2, 1]], dtype=np.float32)
K6 = (1/16)*K6_M
```

# Libreria CV2

```python
B, G, R = cv2.split(img32)

resp_k1_b = cv2.filter2D(B, cv2.CV_32F, K1)
resp_k1_g = cv2.filter2D(G, cv2.CV_32F, K1)
resp_k1_r = cv2.filter2D(R, cv2.CV_32F, K1)
resp_k1_b = cv2.normalize(np.abs(resp_k1_b), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
resp_k1_g = cv2.normalize(np.abs(resp_k1_g), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
resp_k1_r = cv2.normalize(np.abs(resp_k1_r), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
out_k1_bgr = cv2.merge([resp_k1_b, resp_k1_g, resp_k1_r])

resp_k2_b = cv2.filter2D(B, cv2.CV_32F, K2)
resp_k2_g = cv2.filter2D(G, cv2.CV_32F, K2)
resp_k2_r = cv2.filter2D(R, cv2.CV_32F, K2)
resp_k2_b = cv2.normalize(np.abs(resp_k2_b), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
resp_k2_g = cv2.normalize(np.abs(resp_k2_g), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
resp_k2_r = cv2.normalize(np.abs(resp_k2_r), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
out_k2_bgr = cv2.merge([resp_k2_b, resp_k2_g, resp_k2_r])

resp_k3_b = cv2.filter2D(B, cv2.CV_32F, K3)
resp_k3_g = cv2.filter2D(G, cv2.CV_32F, K3)
resp_k3_r = cv2.filter2D(R, cv2.CV_32F, K3)
resp_k3_b = cv2.normalize(np.abs(resp_k3_b), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
resp_k3_g = cv2.normalize(np.abs(resp_k3_g), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
resp_k3_r = cv2.normalize(np.abs(resp_k3_r), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
out_k3_bgr = cv2.merge([resp_k3_b, resp_k3_g, resp_k3_r])

resp_k4_b = cv2.filter2D(B, cv2.CV_32F, K4)
resp_k4_g = cv2.filter2D(G, cv2.CV_32F, K4)
resp_k4_r = cv2.filter2D(R, cv2.CV_32F, K4)
resp_k4_b = cv2.normalize(np.abs(resp_k4_b), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
resp_k4_g = cv2.normalize(np.abs(resp_k4_g), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
resp_k4_r = cv2.normalize(np.abs(resp_k4_r), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
out_k4_bgr = cv2.merge([resp_k4_b, resp_k4_g, resp_k4_r])

resp_k5_b = cv2.filter2D(B, cv2.CV_32F, K5)
resp_k5_g = cv2.filter2D(G, cv2.CV_32F, K5)
resp_k5_r = cv2.filter2D(R, cv2.CV_32F, K5)
resp_k5_b = cv2.normalize(np.abs(resp_k5_b), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
resp_k5_g = cv2.normalize(np.abs(resp_k5_g), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
resp_k5_r = cv2.normalize(np.abs(resp_k5_r), None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
out_k5_bgr = cv2.merge([resp_k5_b, resp_k5_g, resp_k5_r])
```

```python
plt.figure(figsize=(12, 8))
plt.subplot(2,3,1); plt.imshow(cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)); plt.title("Original")
plt.subplot(2,3,2); plt.imshow(cv2.cvtColor(out_k1_bgr, cv2.COLOR_BGR2RGB)); plt.title("K1")
plt.subplot(2,3,3); plt.imshow(cv2.cvtColor(out_k2_bgr, cv2.COLOR_BGR2RGB)); plt.title("K2")
plt.subplot(2,3,4); plt.imshow(cv2.cvtColor(out_k3_bgr, cv2.COLOR_BGR2RGB)); plt.title("K3")
plt.subplot(2,3,5); plt.imshow(cv2.cvtColor(out_k4_bgr, cv2.COLOR_BGR2RGB)); plt.title("K4")
plt.subplot(2,3,6); plt.imshow(cv2.cvtColor(out_k5_bgr, cv2.COLOR_BGR2RGB)); plt.title("K5")
plt.tight_layout(); plt.show()
```

# Manual

```python
img = Image.open(RUTA).convert("RGB")
img_np = np.array(img, dtype=np.float32)  
alt, anc, c = img_np.shape # (alto, ancho, 3)
```

```python
kh, kw = K3.shape
pad_h, pad_w = kh // 2, kw // 2

# Padding
padded = np.pad(img_np, ((pad_h, pad_h), (pad_w, pad_w), (0,0)), mode='reflect')

# Salida
out = np.zeros_like(img_np)
```

```python
for y in range(alt):
    for x in range(anc):
        for ch in range(c):  # B, G, R
            region = padded[y:y+kh, x:x+kw, ch]  # región de la imagen del tamaño del kernel
            out[y, x, ch] = np.sum(region * K3)  # suma ponderada

# --- Post-procesar (valor absoluto + normalizar a 0-255) ---
out = np.abs(out)
out = (out - out.min()) / (out.max() - out.min()) * 255
out = out.astype(np.uint8)
```

```python
plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(out)
plt.title("Convolución K3")
plt.axis("off")
plt.show()
```

```python
def conv(n):
    if n == 1:
        K = K1
    elif n==2:
        K = K2
    elif n==3:
        K = K3
    elif n==4:
        K = K4
    elif n==5:
        K = K5
    elif n==6:
        K = K6
    kh, kw = K.shape
    pad_h, pad_w = kh // 2, kw // 2

    # Padding
    padded = np.pad(img_np, ((pad_h, pad_h), (pad_w, pad_w), (0,0)), mode='reflect')

    # Salida
    out = np.zeros_like(img_np)
    for y in range(alt):
        for x in range(anc):
            for ch in range(c):  # B, G, R
                region = padded[y:y+kh, x:x+kw, ch]  # región de la imagen del tamaño del kernel
                out[y, x, ch] = np.sum(region * K)  # suma ponderada

    # --- Post-procesar (valor absoluto + normalizar a 0-255) ---
    out = np.abs(out)
    out = (out - out.min()) / (out.max() - out.min()) * 255
    out = out.astype(np.uint8)

    plt.subplot(1,2,1)
    plt.imshow(img)
    plt.title("Original")
    plt.axis("off")

    plt.subplot(1,2,2)
    plt.imshow(out)
    plt.title("Convolución K"+ str(n))
    plt.axis("off")
    plt.show()
    return
```

```python
conv(1)
conv(2)
conv(3)
conv(4)
conv(5)
conv(6)
```
