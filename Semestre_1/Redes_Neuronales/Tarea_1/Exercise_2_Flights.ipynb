We start loading libraries

```python
import pandas as pd
import numpy as np
```

We read the data with help from pandas as pd

```python
flights = pd.read_csv('nycflights13.csv')
flights.shape
flights.head()
```

### 1. Rename Columns

To change column names we create a list with all the columns and their new name to have all items at hand.
Then with rename we take our list of new names

```python
new_names = {
    'year': 'Year',
    'month': 'Month',
    'day': 'Day',
    'dep_time': 'Departure_Time',
    'dep_delay': 'Departure_Delay',
    'arr_time': 'Arrival_Time',
    'arr_delay': 'Arrival_Delay',
    'carrier': 'Airline',
    'tailnum': 'Tail_Number',
    'flight': 'Flight_Number',
    'origin': 'Origin',
    'dest': 'Destination',
    'air_time': 'Flight_Time',
    'distance': 'Distance',
    'hour': 'Hour',
    'minute': 'Minute'
}
flights = flights.rename(columns=new_names)
```

### 2. Filter the flights from and to JFK (or any other airport)

For this we check the columns "Origin" and "Destination" if the register mention "JFK" then we save the information in the table "flights_JFK"

```python
flights_JFK = flights[(flights['Origin'] == 'JFK') | (flights['Destination'] == 'JFK')]
flights_JFK.shape
flights_JFK.head()
```

For this we create a column named "Category" in where we will add the type of flight Outgoing/Incoming.

If Origin or Destination match with "JFK" then will be asigned the proper category.

After that we concat both df and we sort the registers to group them.

```python
# Outocomings
out_flights = flights_JFK[flights_JFK['Origin'] == 'JFK']
out_flights['Category'] = 'Outgoing'  # Add Outgoing to results

# Incomings
in_flights = flights_JFK[flights_JFK['Destination'] == 'JFK']
in_flights['Category'] = 'Incoming'  # Add Incoming to results

# Concat df
combined_df = pd.concat([out_flights, in_flights])

# Sort by category
combined_df.set_index(['Category', combined_df.groupby('Category').cumcount()], inplace=True)

combined_df.shape
combined_df.head()
```

### 3. In each category reorder the flights based on delay time in Ascending order (from smallest to greatest)

With help from sort_values we use the "Departure Delay" column to create a new df with sorted items, then we repeat last process to concat the df from outcoming and incomings.

```python

out_flights_delay = out_flights.sort_values(by =['Departure_Delay'], ascending=True) #ascending=True for smallest to gratest
in_flights_delay = in_flights.sort_values(by =['Departure_Delay'], ascending=True)   #         =False for greatest to smallest

combined_df_delay = pd.concat([out_flights_delay, in_flights_delay])
combined_df_delay.set_index(['Category', combined_df_delay.groupby('Category').cumcount()], inplace=True)

combined_df_delay.shape
combined_df_delay
```
