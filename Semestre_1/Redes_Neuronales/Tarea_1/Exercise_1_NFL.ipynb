For start we load the proper libraries

```python
# import modules
import numpy as np
import pandas as pd
import random
```

Then we load the DB that will be used for this exercise with the help of pandas as "pd"

```python
df = pd.read_csv('NFL.csv')
```

We take the column "Age"

*   fillna will fill missing values
*   mode() calcules the mode from column, in the case that they were more than one then de [0] will take the first mode.
*   astype cast the result as int.

At last we print the data frame to ensure results are the expected.





```python
df['Age'] = df['Age'].fillna(df['Age'].mode()[0]).astype(int)
df
```

We split the Birthday in an array by "/" named "date" this will give us an array of 3 elements in order day, month, year.

```python
date = df['Birthday'][3].split('/')
num = df['Age'][3]+int(date[2])
```

With avg we calculate the average age so we can estimate the current year when DB were collected.
With this data we can estimate a average year of birth.

```python
avg = int(df['Age'].mean())
prom_y = int(num-avg)
prom_y
```

Then as we only chose random day and moth

```python
day = np.random.randint(1,30)
month = np.random.randint(1,12)
prom_bd = f'{day}/{month}/{prom_y}'
prom_bd
```

Then we fill the NaN values in Birthday with this date

```python
df['Birthday'] = df['Birthday'].fillna(prom_bd)
df
```

For "College" as a calitative data we collect the unique values and added to a list named "coll_r".

Then with help from lambda function we chose a random value from our list and asigned to the missing values if the register is not NA/Missing then we left the content as it is.

```python
coll_r = df['College'].unique().tolist()
df['College'] = df['College'].apply(lambda x: random.choice(coll_r) if pd.isna(x) else x)
df
```

Here we estimate the mode and fill missing values with the most common result.

```python
df['Current Status'] = df['Current Status'].fillna(df['Current Status'].mode()[0])
df['Current Team'] = df['Current Team'].fillna(df['Current Team'].mode()[0])
df['Experience'] = df['Experience'].fillna(df['Experience'].mode()[0])
df
```

Here we calculate the mean of the values that we already have, then we fill missing values with the one we calculate

```python
Hw = int(df['Height (inches)'].mean())
df['Height (inches)'] = df['Height (inches)'].fillna(Hw)
df
```

High School is filled with the most common value from the ones we got that is "St. Thomas Aquinas HS"

```python
HS_mode = df['High School'].mode()[0]
print(HS_mode)
df['High School'] = df['High School'].fillna(HS_mode)
df
```

For "High School Location" we face a litte issue as we cant simply check for most common result as this cant be random.

So we check from past column we fill if "St. Thomas Aquinas HS" was assigned then we asign the proper location that is "Fort Lauderdale, FL".

```python
df.loc[df['High School'] == 'St. Thomas Aquinas HS', 'High School Location'] = 'Fort Lauderdale, FL'
df
```

For the rest of the columns we estimate the most common value and fill the missing values. The only special case is number that we cast as int.


```python
df['Name'] = df['Name'].fillna(df['Name'].mode()[0])
df['Number'] = df['Number'].fillna(df['Number'].mode()[0]).astype(int)
df['Player Id'] = df['Player Id'].fillna(df['Player Id'].mode()[0])
df['Position'] = df['Position'].fillna(df['Position'].mode()[0])
df['Weight (lbs)'] = df['Weight (lbs)'].fillna(df['Weight (lbs)'].mode()[0])
df['Years Played'] = df['Years Played'].fillna(df['Years Played'].mode()[0])
df
```
